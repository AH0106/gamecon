### Lista vs conjunto: la diferencia central del proyecto

Jaccard necesita **conjuntos**, porque mide cuántos elementos comparten dos colecciones **sin
contar repetidos**.

````python
lista = ["accion", "accion", "rol"]

print("lista:          ", lista,        "→ len =", len(lista))
print("set(lista):     ", set(lista),   "→ len =", len(set(lista)))
print()
print("Un stand con 40 animes tendría 'action' repetido 40 veces.")
print("El conjunto deja una sola copia, que es lo que Jaccard necesita.")
````

## 1.4 · La indentación ES la sintaxis

En Python no hay llaves. Lo que está «dentro» de un `if`, un `for` o una función se marca con
**cuatro espacios** de sangría. Cambiar la sangría cambia el significado.

````python
print("--- print FUERA del for (una vez) ---")
total = 0
for x in [1, 2, 3]:
    total = total + x      # DENTRO del for: se ejecuta 3 veces
print(total)               # FUERA del for: se ejecuta 1 vez

print("\n--- print DENTRO del for (tres veces) ---")
total = 0
for x in [1, 2, 3]:
    total = total + x
    print(total)           # DENTRO: imprime 1, luego 3, luego 6
````

> **⚠ TRAMPA — Nunca mezclar tabulaciones con espacios.**
> Se ve igual en la pantalla y Python lanza `TabError: inconsistent use of tabs and spaces`.
> Configurar el editor para que la tecla Tab inserte 4 espacios.

## 1.6 · Manejar errores y afirmar

````python
def puede_fallar(x):
    return 10 / x

try:
    r = puede_fallar(0)
except ZeroDivisionError:          # solo si el error es de ese tipo
    print("división por cero: usando plan B")
    r = None
except Exception as e:             # cualquier otro error
    print("falló:", type(e).__name__)
    r = None

print("r =", r)
print()
print("type(e).__name__ da el nombre del error como texto.")
print("Es más legible que imprimir el objeto completo.")
````

````python
total = 6

# Si la condición es True, no pasa nada.
assert total == 6, f"el total no cuadra: era {total}"
print("✓ el assert pasó sin hacer nada")

# Si es False, el programa SE DETIENE con ese mensaje.
try:
    assert total == 99, f"el total no cuadra: era {total}"
except AssertionError as e:
    print("✗ el assert se disparó:", e)

print()
print("El assert es la herramienta central del Paso 3.")
print("El MENSAJE es lo que importa: debe decir qué se rompió y cuánto.")
````

## 1.7 · Funciones: la pieza más importante

Una función es una máquina: **entra** algo por los paréntesis, **sale** algo por el `return`.

````python
def jaccard(a, b):                      # def, nombre, parámetros, dos puntos
    comunes = a & b                     # cuerpo, con 4 espacios de sangría
    totales = a | b
    return len(comunes) / len(totales)  # lo que la función ENTREGA

x = jaccard({"accion", "rol"}, {"accion", "terror"})
print("jaccard =", x)
print("redondeado:", round(x, 3))
````

Las cuatro partes, nombradas:

| Parte | En el ejemplo | Qué es |
|---|---|---|
| Nombre | `jaccard` | Cómo se llama a la función |
| Parámetros | `a`, `b` | Nombres internos de lo que entra |
| Cuerpo | las 3 líneas sangradas | Lo que hace |
| Valor de retorno | `len(comunes)/len(totales)` | Lo que sale |

> **⚠ TRAMPA — Una función sin `return` devuelve `None`.**
> Si una función imprime el resultado en vez de devolverlo, el valor no se puede usar después.
> Es el error de diseño más común.

````python
def contar_mal(datos):
    print(len(datos))          # ✗ se ve en pantalla y se PIERDE

def contar_bien(datos):
    return len(datos)          # ✓ se puede usar después

print("--- la versión MAL ---")
n = contar_mal([1, 2, 3])
print("n vale:", n, "→ tipo:", type(n).__name__)
try:
    print(n + 1)
except TypeError as e:
    print("y al usarlo:", type(e).__name__, "-", e)

print("\n--- la versión BIEN ---")
n = contar_bien([1, 2, 3])
print("n vale:", n, "→ n + 1 =", n + 1)
````

## 1.8 · `lambda`: una función sin nombre

`lambda` es una forma corta de escribir una función de una sola línea, para pasarla como
argumento a otra función. Aparece en todo el proyecto dentro de `.map()` y `.agg()`.

````python
import pandas as pd

s = pd.Series([["a","b","c"], ["d"], []])

# Estas dos son equivalentes:
def largo(xs):
    return len(xs)

print("con def:   ", s.map(largo).tolist())
print("con lambda:", s.map(lambda xs: len(xs)).tolist())

print()
print("Se lee: 'lambda, recibe xs, devuelve len(xs)'.")
print("Los dos puntos separan el parámetro del resultado,")
print("y el return está implícito.")
````

````python
# ejemplos reales de este proyecto
f1 = lambda xs: sorted(set(xs))                     # lista → conjunto ordenado
f2 = lambda t: t.strip().lower()                    # texto → limpio
f3 = lambda t: t if t else ["sin-clasificar"]       # lista vacía → centinela
VOCAB = {"accion", "rol"}
f4 = lambda xs: [x for x in xs if x in VOCAB]       # filtrar una lista

print("f1(['b','a','b']) =", f1(["b","a","b"]))
print("f2('  Marvel  ')  =", repr(f2("  Marvel  ")))
print("f3([])            =", f3([]))
print("f3(['accion'])    =", f3(["accion"]))
print("f4(['accion','x']) =", f4(["accion","x"]))
````

## 1.9 · Comprensiones: construir listas en una línea

Una comprensión es un `for` escrito en una línea que **produce** una lista, un conjunto o un
diccionario.

````python
lista = ["Accion", "ROL", "terror"]

# Estas dos son equivalentes:
resultado = []
for x in lista:
    resultado.append(x.lower())
print("con for:        ", resultado)

print("con comprensión:", [x.lower() for x in lista])

print()
print("Se lee: 'para cada x en lista, pon x.lower()'")
````

````python
lista = ["accion", "", "rol", ""]

print("con filtro (el if va al final):")
print("  ", [x for x in lista if x != ""])

print("\nproduciendo un CONJUNTO (llaves en vez de corchetes):")
print("  ", {x for x in lista if x})

print("\nproduciendo un DICCIONARIO (clave: valor):")
dicc = {"a": "x", "b": "y"}
print("  ", {k: v.upper() for k, v in dicc.items()})

print("\nDOS for: 'aplanar' una lista de listas.")
print("Se leen de IZQUIERDA A DERECHA, igual que los for anidados:")
lista_de_listas = [["a","b"], ["c"], ["a"]]
print("  ", [t for sublista in lista_de_listas for t in sublista])
````

## 1.13 · Cómo leer un error

Cuando Python falla, imprime un **traceback**. Se lee **de abajo hacia arriba**: la última
línea dice qué pasó, y las de arriba dicen por dónde pasó el programa para llegar ahí.

```
Traceback (most recent call last):
  File "main.py", line 12, in <module>          ← 4º: empezó aquí
    principal()
  File "main.py", line 8, in principal          ← 3º: luego aquí
    stands, inf = generar_stands(juegos)
  File "src/procesar_tags.py", line 31, in generar_stands
    tags = sorted(grupo["genero"])              ← 2º: LA LÍNEA CULPABLE
TypeError: '<' not supported between instances of 'float' and 'str'
↑ 1º: LEER ESTO PRIMERO
```

**Los dos datos que importan:** la **última línea** (tipo de error y mensaje) y la **última
referencia a un archivo propio** (archivo y número de línea). Si el traceback termina dentro de
un archivo de pandas, hay que subir hasta la última línea que mencione un archivo del
proyecto: ahí está la causa.

En ese ejemplo: se intentó ordenar una columna que mezcla números y texto. La solución no está
en `sorted` sino en el tipo de la columna. Reproducido:

````python
# El error del traceback de arriba, reproducido
mezclada = [2001, "2010", 1999]

try:
    print(sorted(mezclada))
except TypeError as e:
    print("✗", type(e).__name__, "-", e)

print()
print("La causa no es sorted(): es que la columna mezcla int y str.")
print("El arreglo está en el TIPO, no en la operación:")
import pandas as pd
s = pd.Series(mezclada)
print("  pd.to_numeric(s, errors='coerce') →", pd.to_numeric(s, errors="coerce").tolist())
print("  y ahora sí se puede ordenar:      ", sorted(pd.to_numeric(s, errors="coerce")))
````

> **💡 CLAVE — El procedimiento para desatascarse, en orden:**
> 1. Leer la **última línea** del traceback.
> 2. Encontrar la **última línea de un archivo propio**.
> 3. Imprimir el **valor y el tipo** de las variables de esa línea: `print(x)` y
>    `print(type(x))`.
> 4. Buscar el texto del error en el **catálogo de la subsección 5.3**.
>
> Nueve de cada diez veces el problema es que una variable no es del tipo que se creía: un
> texto donde se esperaba un número, una lista donde se esperaba un conjunto, un `NaN` donde
> se esperaba un nulo.

### `pd.read_json(ruta, lines=True, nrows=None, chunksize=None)`

Con `lines=True` lee **JSON Lines**: un objeto por línea. Es el formato del archivo de

**Sale:** un DataFrame — **salvo** si se pasa `chunksize`, en cuyo caso devuelve un

La compresión `.gz` se detecta sola. Sin `nrows` ni `chunksize`, ese archivo son **89.411

````python
import pandas as pd, json, tempfile, pathlib
from collections import Counter

# simulamos un archivo JSON Lines grande
TMP = pathlib.Path(tempfile.mkdtemp())
ruta = TMP / "demo.jsonl"
with open(ruta, "w", encoding="utf-8") as f:
    for i in range(5000):
        f.write(json.dumps({"id": i, "publisher": ["Acme","Zeta",""][i % 3]}) + "\n")

print("--- nrows: solo las primeras N (para explorar) ---")
print(pd.read_json(ruta, lines=True, nrows=3).to_string(index=False))

print("\n--- chunksize: procesar por pedazos, acumulando SOLO el resumen ---")
acum = Counter()
n_leidos = 0
for trozo in pd.read_json(ruta, lines=True, chunksize=1000):
    n_leidos += len(trozo)
    acum.update(trozo["publisher"].tolist())
    # al cerrar la vuelta, `trozo` se libera: la RAM NO crece
print("filas leídas:", n_leidos, "| resumen:", dict(acum))

print("""
⚠ EL ERROR que arruina el chunksize: hacer lista.append(trozo) dentro
  del for y pd.concat(lista) al final. Eso vuelve a cargar todo en
  memoria, y encima fragmentado. Si hace falta el DataFrame completo,
  el chunksize no sirve: hacen falta menos columnas o Parquet.""")
````

````python
print("=" * 62)
print("Quitar tildes con unicodedata (sin lista de reemplazos)")
print("=" * 62)
import unicodedata

def quitar_tildes(x):
    if not isinstance(x, str):          # nulos y números pasan de largo
        return x
    descompuesto = unicodedata.normalize("NFKD", x)   # 'ó' → 'o' + marca
    return "".join(c for c in descompuesto if not unicodedata.combining(c))

for original in ["Récords", "Acción", "Ciencia-Ficción", "mangá", "Niño"]:
    desc = unicodedata.normalize("NFKD", original)
    print(f"{original:<18} descompuesto: {len(desc)} caracteres  →  {quitar_tildes(original)}")

print()
print("NFKD separa cada letra acentuada en dos: letra base + marca.")
print("combining(c) detecta esas marcas para descartarlas.")
print()
print("⚠ Es mejor que .replace('á','a').replace('é','e')... porque")
print("  funciona con CUALQUIER acento sin enumerarlos.")
````

### Etapas del aggregation pipeline

Un *pipeline* es una **lista de etapas**. Cada etapa recibe lo que salió de la anterior.

| Etapa | Qué hace | Equivalente en pandas |
|---|---|---|
| `{"$match": {...}}` | Filtra documentos | `df[mascara]` |
| `{"$unwind": "$tags"}` | Un documento por elemento del array | `df.explode()` |
| `{"$group": {"_id": "$campo", ...}}` | Agrupa y acumula | `df.groupby().agg()` |
| `{"$project": {...}}` | Elige, renombra y **calcula** campos | `df.assign()` |
| `{"$sort": {"campo": -1}}` | Ordena (−1 descendente) | `df.sort_values()` |
| `{"$limit": 5}` | Corta a N | `df.head(N)` |
| `{"$lookup": {...}}` | Join con otra colección. **Devuelve un array** | `df.merge()` |

> **💡 CLAVE — Por qué `$match` va primero, siempre.**
> Cada etapa procesa lo que le llega. Un `$match` al final filtra **después** de haber
> calculado sobre los 2.765 stands; al principio descarta antes de gastar. Y **solo un
> `$match` inicial puede aprovechar un índice**. Es la optimización más importante de todo
> pipeline.

Dentro de un pipeline, `"$campo"` con un dólar significa «el valor de ese campo del
documento». Sin dólar es un texto literal. Y `"$$t"` con dos dólares es la variable que
declara el `as` de un `$filter`.

````python
print("=" * 62)
print("$unwind + $group + $sort — el 'explode + value_counts' de Mongo")
print("=" * 62)

if HAY_MONGO:
    c = db["demo_query"]
    pipeline = [
        {"$unwind": "$tags"},                            # una fila por tag
        {"$group": {"_id": "$tags", "n": {"$sum": 1}}},  # contar por tag
        {"$sort": {"n": -1}},
        {"$limit": 5},
    ]
    for d in c.aggregate(pipeline):
        print(f"  {d['_id']:<10} {d['n']}")
else:
    print("sin Mongo: celda saltada")
````

````python
print("=" * 62)
print("$filter + $in + $size — Jaccard DENTRO de Mongo (portable)")
print("=" * 62)

if HAY_MONGO:
    MIS_INTERESES = ["accion", "rol"]
    N_INT = len(MIS_INTERESES)
    c = db["demo_query"]

    pipeline = [
        # 1. solo stands con tags Y con algún tag en mis intereses
        {"$match": {"tags.0": {"$exists": True},
                    "tags": {"$in": MIS_INTERESES}}},
        # 2. calcular los dos tamaños
        {"$project": {
            "_id": 0, "nombre": 1, "tags": 1,
            "n_comunes": {"$size": {"$filter": {
                "input": "$tags", "as": "t",
                "cond": {"$in": ["$$t", MIS_INTERESES]}}}},
            "n_stand": {"$size": "$tags"},
        }},
        # 3. jaccard = comunes / (n_stand + n_int - comunes)
        #    porque |A ∪ B| = |A| + |B| - |A ∩ B|
        {"$project": {
            "nombre": 1, "tags": 1, "n_comunes": 1,
            "jaccard": {"$divide": [
                "$n_comunes",
                {"$subtract": [{"$add": ["$n_stand", N_INT]}, "$n_comunes"]}]},
        }},
        {"$sort": {"jaccard": -1}},
        {"$limit": 5},
    ]

    print(f"intereses del asistente: {MIS_INTERESES}\n")
    for d in c.aggregate(pipeline):
        print(f"  {d['nombre']}  tags={d['tags']}  comunes={d['n_comunes']}  "
              f"jaccard={d['jaccard']:.3f}")

    print("""
Verificación a mano: el stand A tiene ["accion","rol"], idéntico a
mis intereses → comunes 2, unión 2, jaccard = 1,000.
El stand B tiene ["accion"] → comunes 1, unión 2, jaccard = 0,500.

⚠ No usar $setIntersection, $mergeObjects, $round ni $text:
  mongomock NO los implementa y el proyecto dejaría de correr
  sin servidor. Esta versión es portable.""")
else:
    print("sin Mongo: celda saltada")
````

Y en `procesar_tags.py`, cambiar `groupby("productor")` por `groupby("productor_clave")`, y

**Por qué dos columnas:** `productor_clave` sirve para agrupar; `productor` es lo que el
asistente lee en el mapa de la feria. No conviene mostrarle «marvel» en minúsculas.

El mecanismo, funcionando:

````python
import pandas as pd, unicodedata

# ── las dos funciones del pseudocódigo, implementadas ──
def vacio_a_nulo(s):
    limpia = s.astype("string").str.strip()
    basura = {"", "nan", "none", "null", "n/a", "-"}
    return limpia.mask(limpia.str.lower().isin(basura))

def quitar_tildes(x):
    if not isinstance(x, str):
        return x
    d = unicodedata.normalize("NFKD", x)
    return "".join(c for c in d if not unicodedata.combining(c))

def clave_de_agrupacion(s):
    return (s.map(quitar_tildes)
             .str.strip()
             .str.lower()
             .str.replace(r"\s+", " ", regex=True))

# ── un ejemplo con los mismos defectos que los datos reales ──
df = pd.DataFrame({"productor": [
    " Marvel ", "marvel", "MARVEL  Comics", "Norma Récords",
    "norma records", "", "   ", None, "Zeta Co",
]})

df["productor"] = vacio_a_nulo(df["productor"])
df["productor_clave"] = clave_de_agrupacion(df["productor"])
display(df)

antes = len(df)
n_sin_productor = int(df["productor_clave"].isna().sum())
df_limpio = df[df["productor_clave"].notna()]

print(f"escrituras distintas ANTES : {df['productor'].nunique()}")
print(f"claves distintas DESPUÉS   : {df_limpio['productor_clave'].nunique()}")
print(f"filas rechazadas (sin productor): {n_sin_productor} de {antes}"
      f"  = {n_sin_productor/antes:.0%}")
print()
print("Las 3 formas de 'Marvel' colapsaron en 2 claves")
print("('marvel' y 'marvel comics' son distintos: eso es correcto),")
print("'Norma Récords' y 'norma records' colapsaron en 1,")
print("y los 3 vacíos salieron como filas RECHAZADAS Y CONTADAS,")
print("no como stands fantasma.")
````

````python
import pandas as pd

def texto_a_lista(v):
    if pd.isna(v) or str(v).strip() == "":
        return []
    return [t.strip().lower() for t in str(v).split(",") if t.strip()]

# un grupo de anime, con las dos columnas
grupo = pd.DataFrame({
    "name":   ["Anime A", "Anime B", "Anime C"],
    "genero": ["Action, Adventure", "Action", "Drama"],
    "Themes": ["Mecha, School", "", "Historical"],
})
display(grupo)

# --- solo genero (lo que hace el código hoy) ---
solo_gen = set()
for _, fila in grupo.iterrows():
    solo_gen.update(texto_a_lista(fila["genero"]))

# --- genero + Themes (el cambio del Paso 2) ---
con_temas = set()
for _, fila in grupo.iterrows():
    con_temas.update(texto_a_lista(fila["genero"]))
    con_temas.update(texto_a_lista(fila["Themes"]))

print("solo 'genero'      :", sorted(solo_gen), f"→ {len(solo_gen)} tags")
print("genero + 'Themes'  :", sorted(con_temas), f"→ {len(con_temas)} tags")
print()
print("⚠ sorted(conjunto), NO el conjunto a secas.")
print("  Un conjunto no garantiza el orden entre ejecuciones; si se guardan")
print("  los tags sin ordenar, los documentos que suben a Mongo CAMBIAN")
print("  entre corridas aunque los datos sean idénticos, y eso rompe la")
print("  idempotencia del Paso 5.")
````

### Paso 3 · El informe de rechazos y la ecuación de conservación

**⏱ 60 min · archivo: `procesar_tags.py` · decisiones: 1**

**Este paso es el corazón de la Entrega 2**, y es lo primero que se muestra el viernes. Un ETL
sin informe no está funcionando: está **callado**.

> **💡 CLAVE — La ecuación que hay que poder mostrar:**
> ```
> leídos = cargados + rechazados
> ```
> Sin excepción, y cada rechazo con el **nombre de la regla** que lo rechazó. En la industria

**Herramientas:** `Counter()` · `len()` · `for clave, grupo in df.groupby()` ·
`grupo["col"].iloc[0]` · `s.map(len)` · `s.value_counts()` · `assert` · `df.to_parquet()`

```
FUNCIÓN generar_stands(df_games, df_anime, df_comics)

    informe = contador vacío

    # --- 1. cuántos entraron ---
    informe["leidos_juego"] = tamaño(df_games)
    informe["leidos_anime"] = tamaño(df_anime)
    informe["leidos_comic"] = tamaño(df_comics)
    informe["leidos_total"] = la suma de los tres

    stands = lista vacía

    # --- 2. construir, fuente por fuente ---
    PARA cada (df, dominio) EN [(df_games,"juego"),
                                (df_anime,"anime"),
                                (df_comics,"manga_comic")]

        PARA cada (clave, grupo) EN df agrupado por "productor_clave"

            tags = el conjunto de etiquetas de todo el grupo   # ver Paso 2

            agregar a stands el diccionario:
                clave   = clave
                nombre  = el primer "productor" original del grupo
                dominio = dominio
                n_obras = tamaño(grupo)
                tags    = ordenar(tags)

    informe["stands_construidos"] = tamaño(stands)

    # --- 3. PARTIR, no borrar en silencio ---
    sin_tags    = los stands cuya lista tags está vacía
    publicables = los stands cuya lista tags NO está vacía

    informe["stands_sin_tags"]      = tamaño(sin_tags)
    informe["stands_publicables"]   = tamaño(publicables)
    informe["sin_tags_por_dominio"] = contar sin_tags agrupados por dominio

    # --- 4. LA AFIRMACIÓN ---
    AFIRMAR que informe["stands_construidos"]
             == informe["stands_publicables"] + informe["stands_sin_tags"]
             , con el mensaje: "se perdieron stands en algún filtro"

    devolver publicables, sin_tags, informe
```

**Por qué `AFIRMAR` y no `imprimir`:** un `assert` **detiene el programa** cuando la condición
es falsa. Si el miércoles alguien agrega un filtro y no lo cuenta, el programa se para ahí en
vez de entregar un número silenciosamente equivocado.

````python
from collections import Counter
import pandas as pd

def construir_stands_demo(fuentes):
    """El mecanismo completo del Paso 3, sobre datos pequeños.

    fuentes: lista de (DataFrame, dominio)
    devuelve: (publicables, sin_tags, informe)
    """
    informe = Counter()
    stands = []

    for df, dominio in fuentes:
        informe[f"leidos_{dominio}"] = len(df)
        for clave, grupo in df.groupby("productor_clave"):
            tags = {t for lista in grupo["tags"] for t in lista}
            stands.append({
                "clave":   clave,
                "nombre":  grupo["productor"].iloc[0],   # el original, para mostrar
                "dominio": dominio,
                "n_obras": len(grupo),
                "tags":    sorted(tags),
            })

    informe["leidos_total"]       = sum(len(df) for df, _ in fuentes)
    informe["stands_construidos"] = len(stands)

    # la PARTICIÓN explícita: no se borra en silencio
    sin_tags    = [s for s in stands if len(s["tags"]) == 0]
    publicables = [s for s in stands if len(s["tags"]) > 0]

    informe["stands_sin_tags"]    = len(sin_tags)
    informe["stands_publicables"] = len(publicables)
    informe["sin_tags_por_dominio"] = dict(Counter(s["dominio"] for s in sin_tags))

    # LA AFIRMACIÓN
    assert informe["stands_construidos"] == (informe["stands_publicables"]
                                             + informe["stands_sin_tags"]), \
        (f"se perdieron stands: construidos={informe['stands_construidos']} "
         f"publicables={informe['stands_publicables']} "
         f"sin_tags={informe['stands_sin_tags']}")

    return publicables, sin_tags, informe


# ── dos fuentes de ejemplo. OJO: 'omega' tiene UN productor y SIN tags ──
juegos = pd.DataFrame({
    "productor":       ["Acme", "Acme", "Zeta Co", "Omega"],
    "productor_clave": ["acme", "acme", "zeta co", "omega"],
    "tags":            [["accion"], ["rol"], ["terror","accion"], []],
})
anime = pd.DataFrame({
    "productor":       ["Studio X", "Studio Y"],
    "productor_clave": ["studio x", "studio y"],
    "tags":            [["mecha","accion"], []],
})

pub, sin, inf = construir_stands_demo([(juegos, "juego"), (anime, "anime")])

display(pd.DataFrame(pub))
print("stands SIN tags:", [s["nombre"] for s in sin])
print()
print("── INFORME ──")
for k, v in inf.items():
    print(f"  {k:<24} {v}")
print()
print("── LA ECUACIÓN ──")
print(f"  {inf['stands_construidos']} construidos"
      f" = {inf['stands_publicables']} publicables"
      f" + {inf['stands_sin_tags']} rechazados")
print(f"  ¿cuadra? {inf['stands_construidos'] == inf['stands_publicables'] + inf['stands_sin_tags']}")
````

````python
# Qué pasa cuando alguien agrega un filtro y NO lo cuenta
from collections import Counter

def construir_mal(stands_construidos, publicables, sin_tags_contados):
    inf = Counter(stands_construidos=stands_construidos,
                  stands_publicables=publicables,
                  stands_sin_tags=sin_tags_contados)
    assert inf["stands_construidos"] == inf["stands_publicables"] + inf["stands_sin_tags"], \
        (f"se perdieron stands: construidos={inf['stands_construidos']} "
         f"publicables={inf['stands_publicables']} sin_tags={inf['stands_sin_tags']}")
    return inf

print("caso bueno: 6 = 4 + 2")
print(" ", dict(construir_mal(6, 4, 2)))

print("\ncaso malo: alguien agregó un filtro que botó 1 stand y no lo contó")
try:
    construir_mal(6, 3, 2)
except AssertionError as e:
    print("  ✗ AssertionError:", e)

print("""
✓ Eso es lo que convierte la ecuación en una GARANTÍA.
  El programa se detiene ahí, con los tres números en el mensaje,
  en vez de entregar un informe silenciosamente equivocado.

⚠ RIESGO FRECUENTE: escribir "limpiamos los datos" en el documento
  no vale nada. "De 106.414 obras leídas se construyeron N stands,
  de los cuales M quedaron publicables y K fueron rechazados por
  estas cuatro reglas" es una frase que no se puede discutir.
  Ese párrafo es la parte mejor pagada del informe.""")
````

#### 5b · Convertir la tabla a documentos

BSON **no sabe** qué es `np.int64`, `NaN` ni `NaT`. El `NaN` se inserta como un decimal raro
que **queda en la base sin dar error**; el `NaT` lanza un error que no menciona ni Mongo ni la
columna culpable.

```
FUNCIÓN valor_para_mongo(v)
    SI v es nulo (None, NaN, NaT) ENTONCES devolver NULO
    SI v es un entero de numpy    ENTONCES devolver entero de Python
    SI v es un decimal de numpy   ENTONCES devolver decimal de Python
    SI v es una fecha de pandas   ENTONCES devolver fecha nativa de Python
    SI v es una lista o tupla     ENTONCES
        devolver [valor_para_mongo(x) PARA cada x EN v]   # recursión
    SI v es un diccionario        ENTONCES
        devolver {k: valor_para_mongo(x) PARA cada (k,x) EN v}
    EN OTRO CASO devolver v
```

**El orden de los `SI` importa:** la comprobación de nulo va **primero**, porque `NaT` es una
fecha a efectos de tipo y `NaN` es un decimal. Si se pregunta por el tipo antes que por el
nulo, los nulos se cuelan.

````python
import pandas as pd, numpy as np, json

def valor_para_mongo(v):
    """Convierte UN valor de pandas/numpy a algo que BSON entienda."""
    # el nulo va PRIMERO: NaT es datetime y NaN es float a efectos de tipo
    if v is None:
        return None
    if isinstance(v, (list, tuple)):
        return [valor_para_mongo(x) for x in v]          # recursión
    if isinstance(v, dict):
        return {k: valor_para_mongo(x) for k, x in v.items()}
    if isinstance(v, np.ndarray):
        return [valor_para_mongo(x) for x in v.tolist()]
    if pd.isna(v):
        return None
    if isinstance(v, np.integer):   return int(v)
    if isinstance(v, np.floating):  return float(v)
    if isinstance(v, np.bool_):     return bool(v)
    if isinstance(v, pd.Timestamp): return v.to_pydatetime()
    return v

def tabla_a_documentos(df):
    return [{k: valor_para_mongo(v) for k, v in fila.items()}
            for fila in df.to_dict("records")]

# ── un DataFrame con TODOS los tipos problemáticos ──
sucio = pd.DataFrame({
    "clave":  ["acme", "zeta"],
    "n_obras": pd.array([3, None], dtype="Int64"),
    "precio":  [np.float64(9.99), np.nan],               # NaN
    "fecha":   pd.to_datetime(["2026-11-06", None]),     # NaT en la segunda
    "tags":    [["accion","rol"], []],
    "n":       [np.int64(5), np.int64(7)],               # np.int64
})
display(sucio)
print("tipos originales:", {k: str(v) for k, v in sucio.dtypes.items()})
print()

docs = tabla_a_documentos(sucio)
for d in docs:
    print(d)
print()
print("tipos de Python resultantes:")
print(" ", {k: type(v).__name__ for k, v in docs[0].items()})
print()
print("LA PRUEBA: si json.dumps no falla, BSON lo acepta")
print(" ", json.dumps(docs, default=str)[:100], "…")
print()
print("⚠ El caso NaT es el peor porque el error no ayuda:")
print("  'NaTType does not support utcoffset'. No menciona Mongo,")
print("  ni la columna, ni la fila. Si aparece, el culpable es una")
print("  columna datetime con nulos.")
````

### Paso 6 · El buscador / recomendador

**⏱ 3 h · archivo nuevo: `src/buscar.py` · decisiones: 1**

La parte «búsqueda» de la entrega, y la demo del viernes: dado un asistente con una lista de
intereses, devolver los K stands más parecidos.

```
FUNCIÓN jaccard(conjunto_A, conjunto_B)
    SI A y B están ambos vacíos ENTONCES devolver 0
    comunes = A Y B        # intersección: A & B
    totales = A O B        # unión: A | B
    devolver tamaño(comunes) / tamaño(totales)

FUNCIÓN recomendar(intereses, stands, k)
    intereses = conjunto(intereses)
    resultados = lista vacía
    PARA cada stand EN stands
        SI stand["tags"] está vacío ENTONCES saltar este stand
        puntaje = jaccard(intereses, conjunto(stand["tags"]))
        SI puntaje > 0 ENTONCES
            agregar (stand["nombre"], stand["dominio"], puntaje) a resultados
    ordenar resultados por el tercer elemento, de MAYOR a menor
    devolver los primeros k
```

**Por qué Jaccard y no «cuántos tags comparten»:** contar comunes premia a los stands enormes.
Un stand con 21 tags comparte más con cualquiera que uno con 2, solo por tamaño. Jaccard divide
por la unión, así que mide **proporción de coincidencia**, no cantidad.

````python
def jaccard(a, b):
    a, b = set(a), set(b)
    if not a and not b:
        return 0.0
    return len(a & b) / len(a | b)

def recomendar(intereses, stands, k=5):
    intereses = set(intereses)
    res = []
    for s in stands:
        if not s["tags"]:                    # sin tags: no se puede comparar
            continue
        p = jaccard(intereses, s["tags"])
        if p > 0:
            res.append((s["nombre"], s["dominio"], round(p, 3)))
    res.sort(key=lambda r: r[2], reverse=True)
    return res[:k]

STANDS = [
    {"nombre": "Acme Games",   "dominio": "juego",       "tags": ["accion","rol"]},
    {"nombre": "Zeta Studio",  "dominio": "anime",       "tags": ["accion","mecha","rol"]},
    {"nombre": "Omega Comics", "dominio": "manga_comic", "tags": ["terror","ciencia-ficcion"]},
    {"nombre": "Sur Press",    "dominio": "manga_comic", "tags": ["accion"]},
    {"nombre": "Sin Clasificar","dominio":"juego",       "tags": []},
]

# ── LA VERIFICACIÓN: un caso calculable a mano ──
A = {"accion","rol"}; B = {"accion","rol","terror"}
print("verificación manual:")
print(f"  A={sorted(A)}  B={sorted(B)}")
print(f"  comunes={len(A & B)}  unión={len(A | B)}  →  {len(A&B)}/{len(A|B)} = {len(A&B)/len(A|B):.3f}")
print(f"  jaccard(A,B) = {jaccard(A,B):.3f}   ← debe coincidir")
print()

for intereses in (["accion","rol"], ["terror","ciencia-ficcion"], ["mecha"]):
    print(f"asistente con {intereses}:")
    for nombre, dom, p in recomendar(intereses, STANDS):
        print(f"    {p:.3f}  {nombre:<15} ({dom})")
    print()
print("'Sin Clasificar' NUNCA aparece: su lista de tags está vacía,")
print("así que Jaccard contra cualquier cosa es 0.")
````

````python
# ⚠ LA TRAMPA: el recomendador que se recomienda a sí mismo
import math

def similares_mal(i, stands, k=3):
    """Busca stands parecidos al stand i — SIN excluirlo."""
    ref = set(stands[i]["tags"])
    res = [(s["nombre"], round(jaccard(ref, s["tags"]), 3))
           for s in stands if s["tags"]]
    res.sort(key=lambda r: r[1], reverse=True)
    return res[:k]

def similares_bien(i, stands, k=3):
    """La forma robusta: poner el propio puntaje en -infinito."""
    ref = set(stands[i]["tags"])
    puntajes = []
    for j, s in enumerate(stands):
        if not s["tags"]:
            continue
        p = float("-inf") if j == i else jaccard(ref, s["tags"])
        puntajes.append((s["nombre"], p))
    puntajes.sort(key=lambda r: r[1], reverse=True)
    # descartar los -inf antes de cortar en k: si no, cuando hay menos
    # de k candidatos el propio stand reaparece al final de la lista
    vivos = [(n, p) for n, p in puntajes if p != float("-inf")]
    return [(n, round(p, 3)) for n, p in vivos[:k]]

# dos stands con tags IDÉNTICOS: el caso que rompe "saltarse el primero"
STANDS2 = [
    {"nombre": "Acme (referencia)", "dominio": "juego", "tags": ["accion","rol"]},
    {"nombre": "Clon de Acme",      "dominio": "anime", "tags": ["accion","rol"]},
    {"nombre": "Zeta",              "dominio": "anime", "tags": ["accion","mecha"]},
]

print("buscando stands parecidos a 'Acme (referencia)':")
print("  SIN excluir:", similares_mal(0, STANDS2))
print("     ↑ sale él mismo en primer lugar con 1,000")
print("  CON exclusión por -inf:", similares_bien(0, STANDS2))
print()
print("⚠ Y 'saltarse el primero' NO sirve: 'Clon de Acme' también vale")
print("  1,000, así que con empates se puede descartar un vecino")
print("  legítimo y dejar dentro el stand de referencia.")
print("  Es exactamente el caso de los 372 stands de cómic.")
````

````python
# ⚠ DECISIÓN 7: ¿todas las etiquetas valen igual? El IDF, medido.
import math
from collections import Counter

STANDS3 = [
    {"nombre": "A", "tags": ["accion","rol","cyberpunk"]},
    {"nombre": "B", "tags": ["accion","rol"]},
    {"nombre": "C", "tags": ["accion","terror"]},
    {"nombre": "D", "tags": ["accion","cyberpunk"]},
]
N = len(STANDS3)

# document frequency: en cuántos stands aparece cada etiqueta
df_tag = Counter(t for s in STANDS3 for t in s["tags"])
idf = {t: math.log(N / c) for t, c in df_tag.items()}

print("document frequency:", dict(df_tag))
print("idf:", {k: round(v, 3) for k, v in sorted(idf.items(), key=lambda kv: -kv[1])})
print()
print("⚠ 'accion' está en los 4 stands: idf = log(4/4) = 0")
print("  No aporta NADA a la similitud, y eso es CORRECTO.")
print()

def jaccard_idf(a, b, idf):
    a, b = set(a), set(b)
    inter = sum(idf.get(t, 0.0) for t in a & b)
    union = sum(idf.get(t, 0.0) for t in a | b)
    return inter / union if union > 0 else 0.0

ref = STANDS3[0]["tags"]
print(f"comparando contra A = {ref}\n")
print(f"  {'stand':<6} {'comunes':<22} {'jaccard':>8} {'jaccard+idf':>12}")
for s in STANDS3[1:]:
    com = sorted(set(ref) & set(s["tags"]))
    print(f"  {s['nombre']:<6} {str(com):<22} {jaccard(ref, s['tags']):>8.3f}"
          f" {jaccard_idf(ref, s['tags'], idf):>12.3f}")

print("""
B comparte {accion, rol} y D comparte {accion, cyberpunk}.
Jaccard simple los trata igual (2 comunes, unión 3).
El ponderado descuenta 'accion' (idf=0) en los dos casos, y lo que
queda es comparar 'rol' contra 'cyberpunk'.

⚠ NO esperar que el IDF cambie el ranking SIEMPRE. Con pocas etiquetas
  el rango de idf es estrecho y el efecto es pequeño. Si se implementa,
  MEDIR si cambia el ranking antes de afirmar que mejora: comparar el
  top-5 con y sin IDF y contar en cuántos asistentes cambia el primer
  puesto. Si no alcanza el tiempo, Jaccard simple es defendible — basta
  decir que IDF está planificado y POR QUÉ.""")
````

### Paso 7 · Índices y la evidencia con `explain()`

**⏱ 1,5 h · sacrificable si hay atraso · decisiones: 1**

Un índice es una **estructura ordenada** (un árbol B) que evita recorrer toda la colección. Sin
índice, `COLLSCAN`: lee todos los documentos. Con índice, `IXSCAN`: baja por el árbol en tiempo
logarítmico.

**El costo:** ocupa espacio y **encarece las escrituras**. Por eso no se indexa «todo por si
acaso», y decirlo así demuestra que se entiende el intercambio.

```
PRIMERO escribir las 3 consultas que la aplicación hace DE VERDAD.
        No inventar índices antes de saber qué preguntas se hacen.

   C1: dame los stands que tengan alguno de estos tags    ← el buscador
   C2: dame los stands de este dominio                    ← el mapa de la feria
   C3: dame el stand con esta clave                       ← la ficha

DESPUÉS un índice por consulta:

   C1 → índice MULTIKEY sobre "tags"
        (multikey = índice sobre un array; Mongo indexa
         cada elemento por separado)
   C2 → índice simple sobre "dominio"
   C3 → ya existe: el _id siempre tiene índice. No hacer nada.

LA EVIDENCIA:
   plan_antes   = explain() de C1  → anotar etapa y totalDocsExamined
   crear el índice
   plan_despues = explain() de C1  → anotar etapa y totalDocsExamined

   SI totalDocsExamined no bajó, el índice está mal elegido.
```

El `explain()` está en la subsección 2.11, con la advertencia de que **mongomock no lo
implementa**. Pero el **mecanismo** del índice sí se puede medir sin servidor:

````python
# El mecanismo del índice, medido sin Mongo: scan lineal vs búsqueda binaria
import numpy as np, pandas as pd

def comparaciones_scan(n):
    """Búsqueda lineal, peor caso: el elemento está al final (≡ COLLSCAN)."""
    return n

def comparaciones_binaria(ordenados, objetivo):
    """Búsqueda binaria sobre datos ORDENADOS (≡ IXSCAN).

    Un B-tree hace esto, pero con nodos de muchas claves en vez de dos
    mitades — de ahí que el árbol sea bajo.
    """
    lo, hi, c = 0, len(ordenados), 0
    while lo < hi:
        c += 1
        mid = (lo + hi) // 2
        if ordenados[mid] == objetivo:
            return c
        if ordenados[mid] < objetivo:
            lo = mid + 1
        else:
            hi = mid
    return c

filas = []
for n in (1_000, 10_000, 100_000, 1_000_000):
    datos = list(range(n))
    filas.append({
        "n": n,
        "scan (COLLSCAN)":  comparaciones_scan(n),
        "binaria (IXSCAN)": comparaciones_binaria(datos, n - 1),
        "log2(n) teórico":  round(float(np.log2(n)), 1),
    })
t = pd.DataFrame(filas)
t["factor"] = (t["scan (COLLSCAN)"] / t["binaria (IXSCAN)"]).round(0).astype(int)
display(t)

print(f"""Al pasar de 1.000 a 1.000.000 de documentos el scan necesita 1.000
veces más trabajo, y el índice solo {t.loc[3,'binaria (IXSCAN)']/t.loc[0,'binaria (IXSCAN)']:.1f}× más. Por eso un índice no
"acelera un poco": cambia la CLASE de complejidad.

Y la columna 'log2(n) teórico' calza con las comparaciones medidas,
lo que confirma que es O(log n) y no una impresión.

⚠ Lo que este experimento NO demuestra: que MongoDB use el índice en
  TU consulta. Eso depende del plan que elija el optimizador, y solo
  lo dice explain() en un Mongo real. El mecanismo es universal;
  la aplicación concreta hay que medirla.""")
````

## 6.2 · Fórmulas

```
Jaccard(A,B)    = |A ∩ B| / |A ∪ B|
unión           = |A| + |B| − |A ∩ B|      ← el truco para Mongo
IDF(t)          = log( N / df(t) )          N     = nº de stands
                                            df(t) = en cuántos aparece t
Jaccard con IDF = Σ idf(A∩B) / Σ idf(A∪B)
Ochiai(A,B)     = |A ∩ B| / √(|A|·|B|)     ← el coseno en binario
conservación    = leídos = cargados + rechazados
share por libro = count / suma de counts de ese libro
```

## 6.4 · Componente: qué es / qué no es

| Componente | Es | No es |
|---|---|---|
| **upsert** | actualiza si existe, inserta si no | un `insert` que revisa primero |
| **idempotencia** | N ejecuciones = 1 ejecución | que el código no falle |
| **índice multikey** | índice sobre un array; indexa cada elemento | un índice por elemento |
| **`$unwind`** | un documento por elemento del array | un *join* |
| **`$lookup`** | el *join* documental; devuelve un **array** | gratis — es la etapa más cara |
| **`$match` inicial** | la única etapa que puede usar un índice | intercambiable de posición |
| **esquema flexible** | el esquema lo valida el código | que no haya esquema |
| **replicación** | la misma copia en N nodos; disponibilidad | repartir los datos |
| **sharding** | datos distintos en cada nodo; escala | respaldo |
| **COLLSCAN** | recorrido completo de la colección | un error — a veces es el plan óptimo |
| **IDF** | peso informativo: 0 para lo omnipresente | una medida de calidad de la etiqueta |
| **`errors="coerce"`** | lo inválido pasa a nulo, y se puede contar | ignorar el problema |
| **`dropna`** | elimina `NaN`, `None`, `NaT` | elimina el texto vacío `""` |

## 6.5 · Cómo defender cada decisión

> **«Elegimos [decisión] porque medimos [número concreto], y [término técnico exacto] resuelve
> exactamente eso. Descartamos [alternativa] porque [consecuencia medida].»**

| Versión | Texto | Por qué |
|---|---|---|
| **Insuficiente** | «Usamos upsert para que no se dupliquen los datos.» | describe, no justifica |

## 6.6 · Los términos técnicos exactos

Usar el de la izquierda, nunca el de la derecha. La paráfrasis cuesta puntos aunque el concepto
esté bien entendido.

| Decir esto | No esto |
|---|---|
| **idempotencia** / **upsert** | «que no se duplique» |
| **reconciliación** / ecuación de conservación | «revisar que cuadre» |
| **vocabulario controlado** / tabla de equivalencias | «una lista de géneros» |
| **índice multikey** | «índice sobre la lista» |
| **COLLSCAN** → **IXSCAN** | «se puso más rápido» |
| **regla ESR** (equality, sort, range) | «el orden de los campos» |
| **esquema flexible**, validado en la aplicación | «no tiene esquema» |
| **índice de Jaccard** / **Ochiai** (coseno binario) | «qué tan parecidos son» |
| **document frequency** / **IDF** | «las etiquetas raras pesan más» |
| **write concern** `w="majority"` | «esperar a los otros nodos» |
| **rack awareness** / política de colocación de réplicas | «se reparten» |
| **subutilización del clúster** + **hot spot** | «no se usa bien» |
| **fuga de datos** (*leakage*) | «se contaminó» |
| **documento embebido** vs **referenciado** | «lo pusimos adentro» |

## 6.7 · ¿Por qué MongoDB y no PostgreSQL?

**Lo que se gana:** **esquema flexible** — las tres fuentes tienen campos distintos y no se
quiere un `ALTER TABLE` por fuente. **Documento embebido** — el stand y su array de etiquetas
se leen juntos, así el *join* desaparece de la consulta más frecuente. **Índice multikey** sobre
el array de tags, que indexa cada etiqueta por separado sin necesidad de una tabla puente.

**Lo que se pierde:** transacciones multi-documento cómodas, integridad referencial impuesta
por la base (aquí la valida la aplicación, con los `assert` del Paso 3), y SQL para consultas
ad hoc.

**Admitir lo segundo es lo que hace creíble lo primero.**

## 6.8 · Resumen ejecutivo en siete líneas

1. **Perfilar antes de decidir.** Las reglas de limpieza salen de números medidos.
2. **El error se declara o no existe.** Informe + `assert` + ecuación.
3. **El `count` mide popularidad, no género.** Por eso el umbral falla.
4. **Idempotencia por diseño.** `replace_one` con `upsert`.
5. **Medir antes y después de cada paso.** «No dio error» no es «funcionó».
6. **Una limitación declarada y medida vale más que un arreglo a medias.**
7. **El documento es la nota.** El código es la evidencia.
